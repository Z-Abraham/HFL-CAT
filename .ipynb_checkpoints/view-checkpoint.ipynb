```python
import json
import open3d as o3d
import numpy as np

# 读取预测结果
pred_path = "/root/autodl-tmp/HFL-Net-main/host_folder/ho3d/pred_epoch_40.json"
with open(pred_path, "r") as f:
    xyz_pred_list, verts_pred_list = json.load(f)

# 取第一个样本的预测结果
sample_idx = 0
joints3d = np.array(xyz_pred_list[sample_idx])  # 21个关节的3D坐标
verts3d = np.array(verts_pred_list[sample_idx])  # 约778个网格顶点的3D坐标

# 可视化网格
mesh = o3d.geometry.TriangleMesh()
mesh.vertices = o3d.utility.Vector3dVector(verts3d)
# MANO 模型的面索引可从 mano_layer 中获取（参考 manolayer_ho3d.py 中的 th_faces）
# 假设已获取面索引 faces（形状为 [N, 3]）
# mesh.triangles = o3d.utility.Vector3iVector(faces)
mesh.compute_vertex_normals()

# 可视化关节（用红色球表示）
joints_pcd = o3d.geometry.PointCloud()
joints_pcd.points = o3d.utility.Vector3dVector(joints3d)
joints_pcd.paint_uniform_color([1, 0, 0])  # 红色

# 显示
o3d.visualization.draw_geometries([mesh, joints_pcd])
```
